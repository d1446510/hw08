### 測試一：使用原始資料 (4 人、折價券 100 元)

```python
# 1. 原始資料輸入
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 資料整理與型別轉換
# 去除頭尾空白並轉為小寫
cleaned_event_name = event_name.strip().lower()

# 移除 NT$、逗號與空白，並轉換為浮點數以利後續計算
cleaned_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())

# 將分攤人數轉為整數
num_people = int(people)

# 3. 分步計算費用
# 服務費 = 原始餐費 × 服務費率
service_charge = cleaned_meal_amount * service_rate

# 應付總額 = 原始餐費 + 服務費 - 折價券
total_payment = cleaned_meal_amount + service_charge - discount

# 每人費用 = 應付總額 ÷ 分攤人數
cost_per_person = total_payment / num_people

# 4. 判斷是否超過預算 (直接使用比較運算，結果為布林值)
is_over_budget = cost_per_person > budget_per_person

# 5. 輸出費用摘要
# {:.2f} 代表將浮點數格式化輸出至小數點後兩位
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{cleaned_meal_amount:.2f} 元")
print(f"服務費：{service_charge:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_payment:.2f} 元")
print(f"分攤人數：{num_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
